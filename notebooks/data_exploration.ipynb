```python
import pandas as pd
```

```python
df = pd.read_excel("../data/raw/Online_Retail.xlsx")
df.head()
```

```python
print("*****Dimensions of the datasets*****",df.shape)
print("*****columns of the datasets*****",df.columns.tolist())
```

```python
df.info()
```

```python
print("*****missing values in the datasets*****\n",df.isnull().sum())
print("*****duplicate values in the datasets*****\n",df.duplicated().sum())
```

```python
duplicates = df[df.duplicated(keep=False)]
duplicates.head(20)
```

```python
#delete duplicates
df_clean=df.drop_duplicates(keep='first').copy()
df_clean.head(20)
```

```python
#negative values in quantity 
negative_quantity=df_clean[df_clean['Quantity']<0]
print("negative value in quantity",negative_quantity.shape)
```

```python
negative_quantity.head(20)
```

```python
begin_c=negative_quantity['InvoiceNo'].astype(str).str.startswith("C")
# astype(str)  : convertit en texte (certains numéros sont des nombres)
# str.startswith("C") : True si le numéro commence par C, sinon False
print("Avec 'C' (annulations) :", begin_c.sum())
print("Sans 'C' (à examiner) :", (~begin_c).sum())# ~ = inverse
```

```python
df_cancelled = df_clean[df_clean["InvoiceNo"].astype(str).str.startswith("C")].copy()
df_sales = df_clean[~df_clean["InvoiceNo"].astype(str).str.startswith("C")].copy()
print("Annulations :", len(df_cancelled))
print("Ventes :", len(df_sales))
```

```python
zero_price=df_sales[df_sales['UnitPrice']==0]
print("Lignes avec prix = 0 :", zero_price.shape[0])
zero_price.head(10)
```

```python
df_sales=df_clean[
    (df_clean['Quantity']>0) &
    (df_clean['UnitPrice']>0) &
    (~df_clean["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

print("Lignes avant :", len(df_clean))
print("Ventes propres :", len(df_sales))
print("Lignes exclues :", len(df_clean) - len(df_sales))
df_sales.head(20)
```

```python
df_rejected=df_clean[
    (df_clean['Quantity']<0) |
    (df_clean['UnitPrice']<=0) |
    (df_clean["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

print("Annulations :", len(df_cancelled))
print("Lignes rejetées au total :", len(df_rejected))
print("Rejetées hors annulations :", len(df_rejected) - len(df_cancelled))
```

```python
# --- VALIDATION ---
print("Quantités <= 0 :", (df_sales["Quantity"] <= 0).sum())
print("Prix <= 0 :", (df_sales["UnitPrice"] <= 0).sum())
print("Factures 'C' :", df_sales["InvoiceNo"].astype(str).str.startswith("C").sum())
print("Doublons :", df_sales.duplicated().sum())
```

```python
for d in (df_sales, df_cancelled, df_rejected):
    d["InvoiceNo"] = d["InvoiceNo"].astype(str)
    d["StockCode"] = d["StockCode"].astype(str)
for d in (df_rejected, df_cancelled):
    d["Description"] = d["Description"].astype("string")
```

```python
df_sales.to_parquet("../data/silver_sales.parquet",  index=False)
df_rejected.to_parquet("../data/rejected_sales.parquet", index=False)
df_cancelled.to_parquet("../data/cancelled_sales.parquet", index=False)
print("fichiers sauvegardés dans le dossier data")
```

```python
df_sales=pd.read_parquet("../data/silver_sales.parquet")
df_sales["Revenue"]=df_sales["Quantity"]*df_sales["UnitPrice"]
print("Chiffre d'affaires total :", round(df_sales["Revenue"].sum(), 2))
print("Nombre de commandes :", df_sales["InvoiceNo"].nunique())
print("Panier moyen :", round(df_sales.groupby("InvoiceNo")["Revenue"].sum().mean(), 2))
```

```python
#gold CA par country
gold_country = (df_sales.groupby("Country")["Revenue"].sum().sort_values(ascending=False).reset_index())
gold_country.head(10)
```

```python
#gold CA par description
gold_description=(df_sales.groupby("Description")["Revenue"].sum().sort_values(ascending=False).reset_index())
gold_description.head(10)
```

```python
#gold CA par mois
gold_month=(df_sales.groupby(df_sales["InvoiceDate"].dt.to_period("M"))["Revenue"].sum().sort_values(ascending=False).reset_index())
gold_month.head(10)
```

```python
gold_country.to_parquet("../data/gold_country.parquet", index=False)
gold_description.to_parquet("../data/gold_description.parquet", index=False)
gold_month.to_parquet("../data/gold_month.parquet", index=False)
print("fichiers sauvegardés dans le dossier data")
```

```python
orders_country = (
    df_sales.groupby("Country")["InvoiceNo"]
    .nunique()
    .sort_values(ascending=False)
    .reset_index()
)
orders_country.head(10)
```

```python
non_produits = ["DOTCOM POSTAGE", "POSTAGE", "Manual", "Discount", "CARRIAGE", "Bank Charges"]

gold_products = (
    df_sales[~df_sales["Description"].isin(non_produits)]
    .groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
gold_products
```

```python
gold_products.to_parquet("../data/gold_products.parquet", index=False)
```
